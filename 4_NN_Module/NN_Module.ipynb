```python
## Create model class
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()
    
    def forward(self, features):
        out = self.linear(features)
        out = self.sigmoid(out)

        return out
```

```python
features = torch.randn(10, 5)  # Example input with 10 samples and 5 features
model = Model(features.shape[1])

model(features)
```

```python
model.linear.weight
```

```python
from torchinfo  import summary

summary(model, input_size=(10, 5))
```

## Another NN model

```python
# 5 input features, 3 NN middle layer neurons, 1 output neuron. So 5*3 + 3*1 = 18 parameters(Weight) in total.
#  Also 3 parameters for bias in the middle layer and 1 parameter for bias in the output layer. 
# So total parameters are 18 + 3 + 1 = 22 parameters.

class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear1 = nn.Linear(num_features, 3)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()
    
    def forward(self, features):
        out = self.linear1(features)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)

        return out
```

```python
features = torch.randn(10, 5)  # Example input with 10 samples and 5 features
model = Model(features.shape[1])
model(features)
```

```python
summary(model, input_size=(10, 5))
```

## Improvement

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(num_features, 3),
            nn.ReLU(),
            nn.Linear(3, 1),
            nn.Sigmoid()
        )
    
    def forward(self, features):
        out = self.network(features)

        return out
```

```python
features = torch.randn(10, 5)  # Example input with 10 samples and 5 features
model = Model(features.shape[1])
model(features)
```
